```python
import cv2
import numpy as np
import os
import matplotlib.pyplot as plt
from tensorflow.keras.models import Model, Sequential
from tensorflow.keras.layers import Dense, Dropout, Input, GlobalAveragePooling2D, Flatten, BatchNormalization, Conv2D, MaxPooling2D, GlobalAveragePooling2D
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from sklearn.model_selection import train_test_split

# ====================== 1. CONFIGURATION ====================== #
DATA_ROOT = 'images'  # Dossier principal contenant les sous-dossiers 0-12
IMG_SIZE = (224, 224)  # Taille requise par EfficientNet
NUM_CLASSES = 13  # 13 classes (0-12)
BATCH_SIZE = 8  # Correspond au nombre d'images par dossier

# ====================== 2. CHARGEMENT DES DONNÉES ====================== #
def load_dataset():
    images = []
    labels = []
    
    print("Chargement des images...")
    for label in range(NUM_CLASSES):
        class_dir = os.path.join(DATA_ROOT, str(label))
        if not os.path.exists(class_dir):
            print(f"Avertissement: Dossier {class_dir} introuvable")
            continue
            
        for img_file in os.listdir(class_dir):
            img_path = os.path.join(class_dir, img_file)
            try:
                img = cv2.imread(img_path)
                if img is None:
                    print(f"Image corrompue: {img_path}")
                    continue
                    
                img = cv2.resize(img, IMG_SIZE)
                img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # Conversion BGR vers RGB
                images.append(img)
                labels.append(label)
            except Exception as e:
                print(f"Erreur avec {img_path}: {str(e)}")
    
    if len(images) == 0:
        raise ValueError("Aucune image valide trouvée. Vérifiez votre structure de dossiers.")
    
    print(f"{len(images)} images chargées avec succès")
    return np.array(images), np.array(labels)

# Chargement des données
X, y = load_dataset()
X = X.astype('float32') / 255.0  # Normalisation
y = to_categorical(y, num_classes=NUM_CLASSES)  # One-hot encoding

# Séparation train/test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=0)


# # Callbacks
callbacks = [
    EarlyStopping(monitor='val_loss', patience=7, restore_best_weights=True),
    ReduceLROnPlateau(monitor='val_loss', factor=0.3, patience=3, verbose=1)
]

    
model = Sequential([
        # Bloc 1
        # Conv2D(32, (3, 3), activation='relu', padding='same', input_shape=(IMG_SIZE[0], IMG_SIZE[1], 3)),
        # BatchNormalization(),
        # Conv2D(32, (3, 3), activation='relu', padding='same'),
        # BatchNormalization(),
        # MaxPooling2D((2, 2)),
        # Dropout(0.2),
        
        # # Bloc 2
        # Conv2D(64, (3, 3), activation='relu', padding='same'),
        # BatchNormalization(),
        # Conv2D(64, (3, 3), activation='relu', padding='same'),
        # BatchNormalization(),
        # MaxPooling2D((2, 2)),
        # Dropout(0.3),
        
        # Bloc 3
        Conv2D(128, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        Conv2D(128, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),
        Dropout(0.4),
        
        # Classification
        GlobalAveragePooling2D(),
        Dense(256, activation='relu'),
        BatchNormalization(),
        Dropout(0.5),
        Dense(NUM_CLASSES, activation='softmax')
])
    
model.compile(
    optimizer=Adam(learning_rate=1e-4),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

train_datagen = ImageDataGenerator(
    rescale=1./255,           # Toujours normaliser !
    rotation_range=25,
    width_shift_range=0.15,
    height_shift_range=0.15,
    zoom_range=0.2,
    horizontal_flip=True,
    vertical_flip=False,
    brightness_range=[0.8,1.2],
    shear_range=0.15
)

history2 = model.fit(
    train_datagen.flow(X_train, y_train, batch_size=32),
    epochs=50,
    validation_data=(X_test, y_test),
    callbacks=callbacks,
    verbose=1
)

model.save("model_emotions.h5")
# Visualisation des courbes
plt.figure(figsize=(15, 6))

# Courbe d'accuracy
plt.subplot(1, 2, 1)
plt.plot(history2.history['accuracy'], label='Train Accuracy')
plt.plot(history2.history['val_accuracy'], label='Validation Accuracy')
plt.title('Evolution de la précision', pad=10)
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend()

# Courbe de loss
plt.subplot(1, 2, 2)
plt.plot(history2.history['loss'], label='Train Loss')
plt.plot(history2.history['val_loss'], label='Validation Loss')
plt.title('Evolution de la loss', pad=10)
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend()

plt.tight_layout()
plt.show()


model.fit(
    X_train,                # données d'entraînement (images)
    y_train,                # labels d'entraînement (one-hot ou entiers selon ta perte)
    epochs=10,              # nombre d'epochs (à ajuster selon ton dataset)
    batch_size=32,          # taille du batch classique
    validation_data=(X_test, y_test),  # données de validation (optionnel mais recommandé)
    callbacks=[],           # callbacks éventuels (ex: EarlyStopping, ModelCheckpoint, etc.)
    verbose=1               # affichage du progrès
)

result=model.evaluate(X_test, y_test, verbose=1)
print(f"Perte sur le test: {result[0]:.4f}")
print(f"Précision sur le test: {result[1]:.4f}")

from keras.models import load_model  # Si vous utilisez Keras/TensorFlow

# Charger votre modèle pré-entraîné
model = load_model('model_emotions.h5')  # Remplacez par le chemin de votre modèle

# Dictionnaire des émotions (ajustez selon votre modèle)
emotions = {0: 'Anger', 1: 'Contempt', 2: 'Disgust', 
            3: 'Fear', 4: 'Happy', 5: 'Neutral', 6: 'Sad', 7: 'Surpri'}

# Initialiser la webcam
cap = cv2.VideoCapture(0)

# Détecteur de visages (haar cascade fourni par OpenCV)
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

while True:
    # Capturer une frame
    ret, frame = cap.read()
    if not ret:
        break
    
    # Convertir en niveaux de gris
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    
    # Détecter les visages
    faces = face_cascade.detectMultiScale(gray, 1.3, 5)
    
    for (x, y, w, h) in faces:
        # Extraire la ROI (Region Of Interest)
        roi_gray = gray[y:y+h, x:x+w]
        
        # Pré-traiter l'image pour votre modèle (redimensionnement, normalisation)
        roi_color = frame[y:y+h, x:x+w]
        roi = cv2.resize(roi_color, (48, 48))  # Conserve les 3 canaux
        roi = cv2.cvtColor(roi, cv2.COLOR_BGR2RGB)  # Conversion BGR→RGB (OpenCV utilise BGR par défaut)
        roi = roi.astype('float') / 255.0
        roi = np.expand_dims(roi, axis=0)  # Shape devient (1, 48, 48, 3)
        
        # Prédiction
        preds = model.predict(roi)[0]
        emotion_label = np.argmax(preds)
        emotion = emotions[emotion_label]
        
        
        # Afficher le résultat
        cv2.rectangle(frame, (x, y), (x+w, y+h), (255, 0, 0), 2)
        cv2.putText(frame, emotion, (x, y-10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (36,255,12), 2)
    
    # Afficher le frame
    cv2.imshow('Détection d\'émotions', frame)
    
    # Quitter avec la touche 'q'
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Libérer les ressources
cap.release()
cv2.destroyAllWindows()


```
